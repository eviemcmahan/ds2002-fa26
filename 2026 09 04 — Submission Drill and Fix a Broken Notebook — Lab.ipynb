### How to read an error before you fix it

Python's traceback is read **bottom up**. The last line names the exception and the problem; the line above it points at your code.

```
Traceback (most recent call last):
  File "<stdin>", line 2, in <module>
    print(greeting)
NameError: name 'greeting' is not defined
```

Four exceptions cover most of what you will hit this semester:

| Exception | Usually means |
|---|---|
| `NameError` | The variable was never created, or the cell that created it never ran |
| `TypeError` | You mixed types — text where a number belongs is the classic |
| `KeyError` | That column or dictionary key does not exist. Check spelling and case |
| `FileNotFoundError` | The path is wrong for *this* environment |

### Bug 6 — the answer changes when you re-run it

Run the next cell three times and watch the total move. Then fix it using what you built Wednesday: leave the source data alone and derive a new column from it. Your fix must give the same total on every run.

````python
sales = pd.DataFrame({
    'game': ['NC State', 'Louisville'],
    'units': [1450, 1610],
})

work = sales.copy() # Create a copy to keep sales data untouched
work['units'] = work['units'] * 1.05
print('total units:', round(work['units'].sum(), 2))
````

````python
# Proof — this must print the same number three times
for _ in range(3):
    work = sales.copy()
    work['units'] = work['units'] * 1.05
    print('total units:', round(work['units'].sum(), 2))
    pass
````

**What was wrong:** The units list in the `sales` dataframe was being overwritten and compounded each time it was run. Source data should never be overwritten, it should be derived from.

### Bug 7 — a path from someone else's laptop

This notebook was written on a machine you do not have. Fix it with the pattern from Wednesday's studio: look for the file, and fall back to the inline sample so the notebook still runs for whoever opens it next.

````python
import os
from io import StringIO

SAMPLE = '''game,units\nNC State,1450\nLouisville,1610\n'''

def data_dir():
    """First folder that actually exists, checked in order of likelihood."""
    candidates = [
        'data',                                # running inside the repo
        '../data',
        '/kaggle/input',                       # Kaggle attached dataset
        '/content/drive/MyDrive/ds2002/data',  # Colab with Drive mounted
        '/content',                            # Colab manual upload
    ]
    for path in candidates:
        if os.path.isdir(path):
            return path
    return '.'

def load_csv(filename, sample=SAMPLE):
    path = os.path.join(data_dir(), filename)
    if os.path.exists(path):                    # If the file exists, open it
        with open(path) as f:
            return pd.read_csv(f)
    else:
        return pd.read_csv(StringIO(sample))    # Otherwise, fall back to a small inline sample
    pass

orders = load_csv('/Users/someone/Desktop/gameday_orders.csv')
orders
````

**What was wrong:** The file that was attempting to be uploaded was missing from it's hardcoded path (it was referencing someone else's machine). There should be a fallback so a missing file doesn't kill the notebook.
