### Q2 — Which SKUs and names all mean *strawberry* Pop-Tart?

List them. Then note which Pop-Tart rows are **not** strawberry — those must stay separate, and saying so is part of the answer.

### Q3 — Build the mapping

A dict from each variant SKU to one canonical SKU. Then apply it in a way that leaves unmapped SKUs as themselves rather than turning them into `NaN` — that single mistake silently deletes most of the catalog.

```python
# Strawberry Pop-Tart variants -> one canonical SKU
CANON = {
    'POPTART-STRAW': 'POPTART-STRAW',
    'PT-12': 'POPTART-STRAW',
    'SKU#459812': 'POPTART-STRAW',
    'PT_STRAWBERRY': 'POPTART-STRAW',
}

clean = cat.drop_duplicates()  # dedupe first
clean = clean.assign(canonical=clean['sku'].map(CANON).fillna(clean['sku']))  # unmapped SKUs keep their own value
print('rows:', len(cat), '->', len(clean))
print('distinct SKUs:', clean['sku'].nunique(), '->', clean['canonical'].nunique())
```

### Q4 — Normalize `category`

Strip and title-case, then show the distinct values before and after with counts. Note any pair that is still two labels for one thing — `Breakfast` and `Breakfast Foods` are not the same string but they are the same category.

```python
print("Before:", clean['category'].nunique(), "distinct")
display(clean['category'].value_counts())

clean['category'] = clean['category'].str.strip().str.title()  # remove extra spaces, fix case

print("After strip + title:", clean['category'].nunique(), "distinct")
display(clean['category'].value_counts())

# Remaining duplicate labels -> one category name
CATEGORY_MAP = {
    'Breakfast Foods': 'Breakfast',
    'Emergency Supplies': 'Emergency',
    'Groceries': 'Grocery',
    'Alcoholic Beverages': 'Alcohol',
    'Baby Products': 'Baby',
}

clean['category'] = clean['category'].replace(CATEGORY_MAP)  # unmapped categories stay as-is

print("After merging:", clean['category'].nunique(), "distinct")  # expect 10
display(clean['category'].value_counts())
```

### Q5 — Find the fragmentation you have not spotted yet

Hand-scanning a catalog does not scale. **TODO:** build a candidate list programmatically — normalize each product name (lowercase, strip punctuation, sort the words) and group by that, then show any group holding more than one SKU.

This is the workflow from Monday's lecture, and it is how you will actually approach the full catalog on the midterm.

```python
# Normalize names (lowercase, punctuation to spaces, sort the words)
clean['normalized'] = (cat['product_name'].str.lower()
                      .str.replace(r'[^a-z ]', '', regex=True)
                      .str.split().apply(lambda w: ' '.join(sorted(w))))


candidates = (clean.groupby('normalized')['sku'] # Keep only keys shared by more than one SKU
              .agg(list).loc[lambda s: s.str.len() > 1])
candidates
```
